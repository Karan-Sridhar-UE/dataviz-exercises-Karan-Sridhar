```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Static PNG output so charts show on GitHub (GitHub doesn't run JavaScript)
pio.renderers.default = "png"


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE

asia = df[(df['continent'] == 'Asia') & 
          (df['year'].isin([2002, 2007]))].copy()

# year as string -> categorical colours (2 colours only), not a continuous scale
asia['year'] = asia['year'].astype(str)

fig = px.scatter(
    data_frame=asia,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    hover_name='country',
    log_x=True,
    color_discrete_map={'2002': '#9ECAE1', '2007': '#D62728'},
    category_orders={'year': ['2002', '2007']},
    labels={
        'gdpPercap': 'GDP per Capita (log scale)',
        'lifeExp': 'Life Expectancy (years)',
        'year': 'Year'
    },
    template='plotly_white',
    opacity=0.7
)

fig.update_layout(
    title='Asia: Life Expectancy Increased in Many Countries from 2002 to 2007',
    font=dict(family='Arial', size=13),
    height=600,
    width=1100
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE
import numpy as np

df_2007 = df[df['year'] == 2007].copy()

df_2007['group'] = df_2007['continent'].apply(
    lambda x: 'Asia' if x == 'Asia' else 'Other continents'
)

fig = px.scatter(
    data_frame=df_2007,
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='group',
    hover_name='country',
    log_x=True,
    size_max=60,
    color_discrete_map={
        'Asia': '#D62728',
        'Other continents': 'lightgrey'
    },
    labels={
        'gdpPercap': 'GDP per Capita (log scale)',
        'lifeExp': 'Life Expectancy (years)',
        'pop': 'Population',
        'group': 'Group'
    },
    template='plotly_white',
    title='Asia: Large Population, but Wide Differences in Life Expectancy'
)
# NOTE: on a log axis, annotation x must be given in log10 units
fig.add_annotation(
    x=np.log10(2452), y=64.7,          # India
    ax=-150, ay=-110,                  # text offset in pixels
    text='Asian countries vary widely<br>in wealth and life expectancy',
    showarrow=True,
    arrowhead=2,
    bgcolor='white',
    bordercolor='#D62728'
)
fig.update_layout(
    height=600,
    width=1100,
    font=dict(family='Arial', size=13)
)
fig.show()
```
